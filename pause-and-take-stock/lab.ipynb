## A. Representation and metric

Generate one noisy circle. Compare Euclidean distances before and after an anisotropic coordinate rescaling. Is the topology changing, or the notion of closeness used to construct it?

```python
theta = np.linspace(0, 2*np.pi, 80, endpoint=False)
points = np.c_[np.cos(theta), np.sin(theta)] + 0.06 * RNG.normal(size=(80, 2))
rescaled = points * np.array([2.5, 0.55])

fig, ax = plt.subplots(figsize=(5, 5))
ax.scatter(points[:, 0], points[:, 1], s=16)
ax.set_aspect("equal")
ax.set_title("Original representation")
plt.show()

fig, ax = plt.subplots(figsize=(5, 5))
ax.scatter(rescaled[:, 0], rescaled[:, 1], s=16)
ax.set_aspect("equal")
ax.set_title("Same samples after anisotropic scaling")
plt.show()
```

## C. Filtration direction

Construct a small scalar field containing one peak and one trough. Compare sublevel and superlevel filtrations. Which physical events are being treated as births?

```python
x = np.linspace(-2, 2, 100)
y = np.linspace(-2, 2, 100)
X, Y = np.meshgrid(x, y)
field = np.exp(-((X-0.8)**2 + Y**2)) - 0.8*np.exp(-((X+0.8)**2 + Y**2))
fig, ax = plt.subplots(figsize=(6, 5))
im = ax.imshow(field, origin="lower", extent=[-2,2,-2,2])
fig.colorbar(im, ax=ax)
ax.set_title("One field; two possible filtration directions")
plt.show()
```
